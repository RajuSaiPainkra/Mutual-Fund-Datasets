```python
# import library
import pandas as pd
```

```python
data = pd.read_csv("/kaggle/input/india-mutual-fund-dataset-2025/Mutual Fund Dataset 2025.csv")
```

## Data Exploration and Initilization

```python
data.head()
```

```python
rows, columns = data.shape
print("Rows in the Dataset : \n", rows)
print("Colums in the Datasets : \n", columns)
```

```python
# check Columns 
print("Columns : \n", data.columns)
print ("\nCount of Columns : \n", len(data.columns))
```

```python
# Datasets Column Categorization : 
## Numerical Column
## Categorical Column
## Ordinal Columns

num_column = []
cat_column = []
ord_column = []

for colm in data.columns:
    if data[colm].dtypes == "object":
        cat_column.append(colm)
    elif data[colm].dtypes == "int64" or data[colm].dtypes == "float64":
        num_column.append(colm)
    else:
        ord_column.append(colm)

print("Numerical Columns : \n", num_column)
print("Numerical Columns count : \n", len(num_column))


print("Categorical Columns : \n", cat_column)
print("Categorical Columns Count : \n", len(cat_column))


print("Ordinal Columns : \n", ord_column)
print("Ordinal Columns Count : \n", len(ord_column))



# Check the columns count 

print( "Total Column in Datasets : \n", len(data.columns))
print( " Sum of all Categorize Columns : \n", len(num_column) + len(cat_column) + len(ord_column))
```

```python
# All columns are Categorical Data , means Alphanumerica and Symbol are available in the Datasets column
```

```python
## task 1 : Exploring the Fund Types in the Mutual fund data type
```

```python
fund_type = data["Fund Type"]
```

```python
print("Rows, Columns : \n", fund_type.shape)
```

```python
# checking the unique fund type
print("Count of the Columns unique values : ", len(fund_type.unique()))
print("Unique value in the fund type columns : ", fund_type.unique())
```

```python
# Check How Many Times : 
## Open Ended repeat
open_ended_count = (fund_type == "Open Ended").sum()
print("Count Open Ended : ", open_ended_count)

## Closed Ended repeat
closed_ended_count = (fund_type == "Closed Ended").sum()
print("Count Closed Ended : ", closed_ended_count)

## Interval
interval_count = (fund_type == "Interval").sum()
print("Count Interval : ", interval_count)
```
